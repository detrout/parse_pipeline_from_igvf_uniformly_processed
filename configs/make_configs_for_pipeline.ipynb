```python
def find_config_tsv_files(main_dir):
    return [
        os.path.join(root, file)
        for root, dirs, files in os.walk(main_dir)
        for file in files
        if file.endswith('_config.tsv')
    ]
```

```python
configs = find_config_tsv_files('/share/crsp/lab/seyedam/share/parse_pipeline_from_igvf/configs/configs_fixed_temp')
configs_dict = {
    os.path.basename(path).replace('_config.tsv', ''): path
    for path in configs
}
```

```python
configs_dict
```

```python
plate = 'igvf_017'


h5ad_path = '/share/crsp/lab/seyedam/share/IGVF_portal_h5ads/scripts/h5ad_from_portal_new/' + plate
file_dict = {
    p.stem.split('_')[1]: str(p)
    for p in Path(h5ad_path).iterdir()
    if p.is_file() and '67' in p.name
}


config = pd.read_csv(configs_dict[plate], sep = '\t')

config['igvf_adata'] = config['subpool'].map(file_dict)

if 'Unnamed: 0' in config.columns:
    config.drop('Unnamed: 0', axis = 1, inplace = True)

if 'Unnamed: 0.1' in config.columns:
    config.drop('Unnamed: 0.1', axis = 1, inplace = True)


config.head()
```

```python
config.to_csv(
    f'/share/crsp/lab/seyedam/share/parse_pipeline_from_igvf/configs/{plate}_config.tsv',
    sep='\t',
    index=False
)
```
